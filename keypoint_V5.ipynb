```python
from pathlib import Path
import shutil
import random

base_path = Path("/content/gmas_keypoint")

source_images = base_path / "train" / "images"
source_labels = base_path / "train" / "labels"

# Get all original images
images = sorted([
    p for p in source_images.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
])

print("Total images:", len(images))

# Shuffle consistently
random.seed(42)
random.shuffle(images)

# Split sizes
total = len(images)
train_count = int(total * 0.70)
valid_count = int(total * 0.15)

train_images = images[:train_count]
valid_images = images[train_count:train_count + valid_count]
test_images = images[train_count + valid_count:]

print("Training images:", len(train_images))
print("Validation images:", len(valid_images))
print("Test images:", len(test_images))

# Create validation and test folders
for split_name, split_images in [
    ("valid", valid_images),
    ("test", test_images)
]:

    image_dir = base_path / split_name / "images"
    label_dir = base_path / split_name / "labels"

    image_dir.mkdir(parents=True, exist_ok=True)
    label_dir.mkdir(parents=True, exist_ok=True)

    for image_path in split_images:

        # Copy image
        shutil.copy2(
            image_path,
            image_dir / image_path.name
        )

        # Copy corresponding label
        label_path = source_labels / f"{image_path.stem}.txt"

        if label_path.exists():
            shutil.copy2(
                label_path,
                label_dir / label_path.name
            )
        else:
            print("WARNING: Missing label:", image_path.name)

print("\nSplitting completed.")
```

```python
from pathlib import Path
import shutil

# Original dataset
source_base = Path("/content/gmas_keypoint")

# New YOLO pose dataset
pose_base = Path("/content/gmas_pose")

# Number of keypoints
NUM_KEYPOINTS = 23

# Class names
names = [
    "Crotch",
    "L.AH",
    "L.FBP",
    "L.FFP",
    "L.HP",
    "L.LOE",
    "L.NEP",
    "L.SNP",
    "L.SOE",
    "L.SOS",
    "L.SP",
    "L.SSE",
    "R.AH",
    "R.FBP",
    "R.FFP",
    "R.HP",
    "R.LOE",
    "R.NEP",
    "R.SNP",
    "R.SOE",
    "R.SOS",
    "R.SP",
    "R.SSE"
]

# Process each split
for split in ["train", "valid", "test"]:

    source_images = source_base / split / "images"
    source_labels = source_base / split / "labels"

    output_images = pose_base / split / "images"
    output_labels = pose_base / split / "labels"

    output_images.mkdir(parents=True, exist_ok=True)
    output_labels.mkdir(parents=True, exist_ok=True)

    for image_path in source_images.iterdir():

        # Copy image
        shutil.copy2(
            image_path,
            output_images / image_path.name
        )

        # Original annotation
        label_path = source_labels / f"{image_path.stem}.txt"

        # Store keypoints:
        # class_id -> (x, y)
        keypoints = {}

        if label_path.exists():

            for line in label_path.read_text().strip().splitlines():

                values = line.split()

                if len(values) != 5:
                    continue

                class_id = int(values[0])

                x_center = float(values[1])
                y_center = float(values[2])

                # Only accept valid class IDs
                if 0 <= class_id < NUM_KEYPOINTS:
                    keypoints[class_id] = (
                        x_center,
                        y_center
                    )

        # YOLO pose format:
        # class bbox_x bbox_y bbox_w bbox_h
        # kp1_x kp1_y kp1_visibility
        # ...
        #
        # Since this dataset represents keypoints,
        # create one object covering the whole image.
        bbox_x = 0.5
        bbox_y = 0.5
        bbox_w = 1.0
        bbox_h = 1.0

        output_values = [
            "0",
            str(bbox_x),
            str(bbox_y),
            str(bbox_w),
            str(bbox_h)
        ]

        # Add all 23 keypoints in fixed order
        for class_id in range(NUM_KEYPOINTS):

            if class_id in keypoints:

                x, y = keypoints[class_id]

                # 2 = visible
                output_values.extend([
                    str(x),
                    str(y),
                    "2"
                ])

            else:

                # 0 = missing/not visible
                output_values.extend([
                    "0",
                    "0",
                    "0"
                ])

        # Save converted label
        output_label_path = output_labels / f"{image_path.stem}.txt"

        output_label_path.write_text(
            " ".join(output_values) + "\n"
        )

    print(
        f"{split.upper()} conversion complete: "
        f"{len(list(output_images.iterdir()))} images"
    )

print("\nConversion completed.")
print("New dataset:", pose_base)
```

```python
import gradio as gr
from ultralytics import YOLO
import cv2
import numpy as np

model = YOLO("/content/runs/gmas_23keypoint/weights/best.pt")

keypoint_names = [
    "Crotch",
    "L.AH",
    "L.FBP",
    "L.FFP",
    "L.HP",
    "L.LOE",
    "L.NEP",
    "L.SNP",
    "L.SOE",
    "L.SOS",
    "L.SP",
    "L.SSE",
    "R.AH",
    "R.FBP",
    "R.FFP",
    "R.HP",
    "R.LOE",
    "R.NEP",
    "R.SNP",
    "R.SOE",
    "R.SOS",
    "R.SP",
    "R.SSE"
]

def predict(image):

    if image is None:
        return None

    results = model.predict(
        source=image,
        imgsz=640,
        conf=0.25,
        device=0
    )

    result = results[0]

    # Convert image to OpenCV format
    output = image.copy()

    if result.keypoints is not None and len(result.keypoints.xy) > 0:

        points = result.keypoints.xy[0].cpu().numpy()

        if result.keypoints.conf is not None:
            confs = result.keypoints.conf[0].cpu().numpy()
        else:
            confs = np.ones(len(points))

        for i, (x, y) in enumerate(points):

            # Only display reasonably confident points
            if confs[i] < 0.25:
                continue

            x = int(x)
            y = int(y)

            # Draw a small point
            cv2.circle(
                output,
                (x, y),
                4,
                (0, 0, 0),
                -1
            )

            # Display keypoint name
            label = keypoint_names[i]

            cv2.putText(
                output,
                label,
                (x + 6, y - 6),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45,
                (0, 0, 0),
                2,
                cv2.LINE_AA
            )

            cv2.putText(
                output,
                label,
                (x + 6, y - 6),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45,
                (255, 255, 255),
                1,
                cv2.LINE_AA
            )

    return output


interface = gr.Interface(
    fn=predict,
    inputs=gr.Image(
        type="numpy",
        label="Upload Garment Image"
    ),
    outputs=gr.Image(
        type="numpy",
        label="Detected Keypoints"
    ),
    title="GMAS - 23 Keypoint Detection",
    description="Upload a garment image to detect and label the 23 garment keypoints."
)

interface.launch(share=True)
```

```python
import gradio as gr
from ultralytics import YOLO
import cv2
import numpy as np

# Load trained model
model = YOLO("/content/runs/gmas_23keypoint/weights/best.pt")

# 23 keypoint names
keypoint_names = [
    "Crotch",
    "L.AH",
    "L.FBP",
    "L.FFP",
    "L.HP",
    "L.LOE",
    "L.NEP",
    "L.SNP",
    "L.SOE",
    "L.SOS",
    "L.SP",
    "L.SSE",
    "R.AH",
    "R.FBP",
    "R.FFP",
    "R.HP",
    "R.LOE",
    "R.NEP",
    "R.SNP",
    "R.SOE",
    "R.SOS",
    "R.SP",
    "R.SSE"
]


def predict(image):

    if image is None:
        return None

    # Predict with very low detection confidence
    results = model.predict(
        source=image,
        imgsz=640,
        conf=0.01,
        device=0,
        verbose=False
    )

    result = results[0]

    # Make a copy
    output = image.copy()

    # Check whether keypoints exist
    if result.keypoints is None:
        cv2.putText(
            output,
            "NO KEYPOINTS DETECTED",
            (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )
        return output

    # Get keypoint coordinates
    points = result.keypoints.xy.cpu().numpy()

    # Get confidence values
    if result.keypoints.conf is not None:
        confidences = result.keypoints.conf.cpu().numpy()
    else:
        confidences = np.ones((len(points), 23))

    print("Number of detected objects:", len(points))

    # Process each detected garment
    for obj_index in range(len(points)):

        obj_points = points[obj_index]
        obj_conf = confidences[obj_index]

        print(f"\nObject {obj_index + 1}")

        # Draw ALL 23 keypoints
        for i in range(23):

            x, y = obj_points[i]
            confidence = obj_conf[i]

            x = int(x)
            y = int(y)

            print(
                f"{i:2d}  {keypoint_names[i]:8s} "
                f"x={x:4d} y={y:4d} conf={confidence:.3f}"
            )

            # Draw point
            cv2.circle(
                output,
                (x, y),
                6,
                (0, 0, 255),
                -1
            )

            # Keypoint name + confidence
            label = f"{keypoint_names[i]} ({confidence:.2f})"

            # Black outline
            cv2.putText(
                output,
                label,
                (x + 8, y - 8),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45,
                (0, 0, 0),
                3,
                cv2.LINE_AA
            )

            # White text
            cv2.putText(
                output,
                label,
                (x + 8, y - 8),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45,
                (255, 255, 255),
                1,
                cv2.LINE_AA
            )

    return output


interface = gr.Interface(
    fn=predict,
    inputs=gr.Image(
        type="numpy",
        label="Upload Garment Image"
    ),
    outputs=gr.Image(
        type="numpy",
        label="23 Detected Keypoints"
    ),
    title="GMAS - 23 Keypoint Detection",
    description="All predicted keypoints are displayed with their names and confidence scores."
)

interface.launch(share=True)
```
